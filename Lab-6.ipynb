```python
import gymnasium as gym
import numpy as np


# ============================================================
# RANDOM AGENT
# ============================================================

def random_agent(env_name, num_episodes=10):

    print("=" * 60)
    print(f"ENVIRONMENT: {env_name}")
    print("=" * 60)

    # 1. Create the environment
    env = gym.make(env_name)

    episode_rewards = []

    # Repeat for each episode
    for episode in range(1, num_episodes + 1):

        # 2 & 3. Reset environment and observe initial state
        state, info = env.reset()

        done = False
        total_reward = 0
        steps = 0

        # 4-8. State -> Action -> Reward -> Next State
        while not done:

            # 5. Select a random action
            action = env.action_space.sample()

            # 6. Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # 7. Observe next state and reward
            state = next_state
            total_reward += reward
            steps += 1

            # Episode terminates when either condition occurs
            done = terminated or truncated

        # Store total reward
        episode_rewards.append(total_reward)

        # 9 & 10. Print reward for each episode
        print(
            f"Episode {episode:2d} | "
            f"Steps: {steps:5d} | "
            f"Total Reward: {total_reward:8.0f}"
        )

    # Calculate average reward
    average_reward = np.mean(episode_rewards)

    print("-" * 60)
    print(f"Average Reward: {average_reward:.2f}")
    print("=" * 60)

    env.close()

    return episode_rewards, average_reward


# ============================================================
# 1. CLIFF WALKING
# ============================================================

cliff_rewards, cliff_average = random_agent(
    "CliffWalking-v1",
    num_episodes=10
)


# ============================================================
# 2. TAXI
# ============================================================

taxi_rewards, taxi_average = random_agent(
    "Taxi-v4",
    num_episodes=10
)
```
